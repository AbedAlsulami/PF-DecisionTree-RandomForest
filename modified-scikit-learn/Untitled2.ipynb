```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier

X, y = load_iris(return_X_y=True)

tests = [
    ("gini", {}),
    ("entropy", {}),
    ("entropy_alpha", {"alpha": 0.6}),
    ("gini_beta", {"beta": 0.5}),
    ("proposed", {"alpha": 0.6, "beta": 0.5, "gamma": 0.8}),
]

for criterion, params in tests:
    clf = DecisionTreeClassifier(
        criterion=criterion,
        random_state=42,
        **params
    )
    clf.fit(X, y)

    print("=" * 50)
    print("Criterion:", criterion)
    print("Params:", clf.get_params())
    print("Accuracy:", clf.score(X, y))
    print("Depth:", clf.get_depth())
    print("Leaves:", clf.get_n_leaves())
```

```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import StratifiedKFold, cross_val_score

X, y = load_iris(return_X_y=True)

tests = [
    ("gini", {}),
    ("entropy", {}),
    ("entropy_alpha", {"alpha": 0.6}),
    ("gini_beta", {"beta": 0.5}),
    ("proposed", {"alpha": 0.6, "beta": 0.5, "gamma": 0.8}),
]

cv = StratifiedKFold(
    n_splits=10,
    shuffle=True,
    random_state=42
)

for criterion, params in tests:
    model = DecisionTreeClassifier(
        criterion=criterion,
        random_state=42,
        **params
    )

    scores = cross_val_score(
        model,
        X,
        y,
        cv=cv,
        scoring="accuracy"
    )

    print("=" * 50)
    print("Criterion:", criterion)
    print("Fold scores:", scores)
    print("Mean accuracy:", scores.mean())
    print("Std:", scores.std())
    
```

```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import StratifiedKFold, cross_val_score

X, y = load_iris(return_X_y=True)

tests = [
    ("gini", {}),
    ("entropy", {}),
    ("entropy_alpha", {"alpha": 0.1}),
    ("gini_beta", {"beta": 0.5}),
    ("proposed", {"alpha": 1.0, "beta": 0.5, "gamma": 1.0}),
]

cv = StratifiedKFold(
    n_splits=10,
    shuffle=True,
    random_state=42
)

for criterion, params in tests:
    model = DecisionTreeClassifier(
        criterion=criterion,
        random_state=42,
        **params
    )

    scores = cross_val_score(
        model,
        X,
        y,
        cv=cv,
        scoring="accuracy"
    )

    print("=" * 50)
    print("Criterion:", criterion)
    print("Fold scores:", scores)
    print("Mean accuracy:", scores.mean())
    print("Std:", scores.std())
    
```

```python
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier

X, y = load_iris(return_X_y=True)

rf = RandomForestClassifier(
    n_estimators=5,
    criterion="proposed",
    alpha=0.6,
    beta=0.5,
    gamma=0.8,
    random_state=42,
)

rf.fit(X, y)

print("Accuracy:", rf.score(X, y))

for i, tree in enumerate(rf.estimators_):
    print(f"\nTree {i}")
    print("criterion :", tree.criterion)
    print("alpha     :", tree.alpha)
    print("beta      :", tree.beta)
    print("gamma     :", tree.gamma)
```

```python
from sklearn.datasets import load_iris
from sklearn.ensemble import GradientBoostingClassifier

X, y = load_iris(return_X_y=True)

gb = GradientBoostingClassifier(
    criterion="proposed",
    alpha=0.6,
    beta=0.5,
    gamma=0.8,
    random_state=42,
)

gb.fit(X, y)

print("Accuracy:", gb.score(X, y))

tree = gb.estimators_[0, 0]

print("criterion :", tree.criterion)
print("alpha     :", tree.alpha)
print("beta      :", tree.beta)
print("gamma     :", tree.gamma)
```

```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import AdaBoostClassifier

X, y = load_iris(return_X_y=True)

base = DecisionTreeClassifier(
    criterion="proposed",
    alpha=0.6,
    beta=0.5,
    gamma=0.8,
    max_depth=1,
)

ada = AdaBoostClassifier(
    estimator=base,
    n_estimators=5,
    random_state=42,
)

ada.fit(X, y)

print("Accuracy:", ada.score(X, y))

for i, tree in enumerate(ada.estimators_):
    print(f"\nEstimator {i}")
    print("criterion :", tree.criterion)
    print("alpha     :", tree.alpha)
    print("beta      :", tree.beta)
    print("gamma     :", tree.gamma)
```
